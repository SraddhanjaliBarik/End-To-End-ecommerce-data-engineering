```python
from pyspark.sql.functions import col
cust = spark.table("Ecommerce_Project.Silver_LH.dbo.customers").alias("c")
orders = spark.table("Ecommerce_Project.Silver_LH.dbo.orders").alias("o")
payments = spark.table("Ecommerce_Project.Silver_LH.dbo.payments").alias("p")
support = spark.table("Ecommerce_Project.Silver_LH.dbo.support_tickets").alias("s")
web = spark.table("Ecommerce_Project.Silver_LH.dbo.web_activities").alias("w")

customer360 = (
    cust
    .join(orders, "customer_id", "left")
    .join(payments, "customer_id", "left")
    .join(support, "customer_id", "left")
    .join(web, "customer_id", "left")
    .select(
        "c.customer_id", "c.name", "c.email", "c.gender", "c.dob", "c.location",
        "o.order_id", "o.order_date", col("o.amount").alias("order_amount"), "o.status",
        "p.payment_method", "p.payment_status", col("p.amount").alias("payment_amount"),
        "s.ticket_id", "s.issue_type", "s.ticket_date", "s.resolution_status",
        "w.page_viewed", "w.device_type", "w.session_time"
    )
)
display(customer360)
customer360.write.format("delta").mode("overwrite").saveAsTable("Ecommerce_Project.Gold_LH.dbo.customer360_data")
```
